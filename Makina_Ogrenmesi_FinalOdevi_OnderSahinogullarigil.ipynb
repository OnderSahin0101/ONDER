```python
# ============================================================
# HIGGS DATASET - FINAL ML PROJECT
# Feature Selection + Hyperparameter Optimization
# Nested Cross Validation
# Outer CV: 5-Fold
# Inner CV: 3-Fold
# Models: KNN, Linear SVM, MLP, XGBoost
# ============================================================

!pip install xgboost ucimlrepo -q

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# from ucimlrepo import fetch_ucirepo

from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.preprocessing import MinMaxScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.pipeline import Pipeline
from sklearn.base import clone

from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.calibration import CalibratedClassifierCV
from sklearn.neural_network import MLPClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    roc_curve
)

# ============================================================
# 1) LOAD DATASET
# ============================================================

print("HIGGS Dataset yükleniyor...")

# higgs = fetch_ucirepo(id=280)

# Load from local CSV file
df = pd.read_csv("HIGGS.csv.gz", compression="gzip")

# Assuming the first column is the target and the rest are features
X = df.iloc[:, 1:]
y = df.iloc[:, 0]

# if isinstance(y, pd.DataFrame):
#     y = y.iloc[:, 0]

print("Original X Shape:", X.shape)
print("Original y Shape:", y.shape)

# ============================================================
# 2) RANDOM SAMPLE
# ============================================================

sample_size = 100000

data = X.copy()
data["target"] = y.values

data_sample = data.sample(n=sample_size, random_state=42)

X = data_sample.drop("target", axis=1)
y = data_sample["target"]

print("Sample X Shape:", X.shape)
print("Sample y Shape:", y.shape)

# ============================================================
# 3) OUTLIER ANALYSIS - IQR
# ============================================================

print("\nAykırı değer analizi yapılıyor...")

X_clean = X.copy()
outlier_summary = []

for col in X.columns:
    Q1 = X[col].quantile(0.25)
    Q3 = X[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((X[col] < lower) | (X[col] > upper)).sum()
    rate = outliers / len(X) * 100

    outlier_summary.append([col, outliers, rate])

    X_clean[col] = np.where(X_clean[col] < lower, lower, X_clean[col])
    X_clean[col] = np.where(X_clean[col] > upper, upper, X_clean[col])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=["Feature", "Outlier_Count", "Outlier_Rate_%"]
)

print("\nEn çok aykırı değer içeren özellikler:")
display(outlier_df.sort_values("Outlier_Rate_%", ascending=False).head(15))

# ============================================================
# 4) MODELS
# ============================================================

models = {
    "KNN": KNeighborsClassifier(),

    "SVM": CalibratedClassifierCV(
        estimator=LinearSVC(max_iter=5000, random_state=42),
        cv=3
    ),

    "MLP": MLPClassifier(
        max_iter=300,
        random_state=42,
        early_stopping=True
    ),

    "XGBoost": XGBClassifier(
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )
}

# ============================================================
# 5) PARAMETER GRIDS
# ============================================================

param_grids = {
    "KNN": {
        "feature_selection__k": [15],
        "model__n_neighbors": [3, 5, 7, 9, 11],
        "model__weights": ["uniform", "distance"]
    },

    "SVM": {
        "feature_selection__k": [15],
        "model__estimator__C": [0.1, 1, 10]
    },

    "MLP": {
        "feature_selection__k": [15],
        "model__hidden_layer_sizes": [(50,), (100,)],
        "model__activation": ["relu", "tanh"],
        "model__learning_rate_init": [0.001, 0.01]
    },

    "XGBoost": {
        "feature_selection__k": [15],
        "model__n_estimators": [100, 200],
        "model__max_depth": [3, 5],
        "model__learning_rate": [0.05, 0.1],
        "model__subsample": [0.8, 1.0]
    }
}

# ============================================================
# 6) NESTED CROSS VALIDATION
# ============================================================

outer_cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

inner_cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=42
)

all_results = []
roc_data = {}

X_final = X_clean.copy()
y_final = y.copy()

print("\nNested Cross Validation başlıyor...")

for model_name, model in models.items():

    print("\n====================================================")
    print(f"MODEL: {model_name}")
    print("====================================================")

    fold_no = 1

    for train_idx, test_idx in outer_cv.split(X_final, y_final):

        print(f"\n{model_name} - Outer Fold {fold_no} çalışıyor...")

        X_train = X_final.iloc[train_idx]
        X_test = X_final.iloc[test_idx]

        y_train = y_final.iloc[train_idx]
        y_test = y_final.iloc[test_idx]

        pipeline = Pipeline([
            ("scaler", MinMaxScaler()),
            ("feature_selection", SelectKBest(score_func=f_classif)),
            ("model", clone(model))
        ])

        grid_search = GridSearchCV(
            estimator=pipeline,
            param_grid=param_grids[model_name],
            cv=inner_cv,
            scoring="roc_auc",
            n_jobs=-1,
            verbose=1
        )

        grid_search.fit(X_train, y_train)

        best_model = grid_search.best_estimator_

        y_pred = best_model.predict(X_test)
        y_proba = best_model.predict_proba(X_test)[:, 1]

        accuracy = accuracy_score(y_test, y_pred)
        precision = precision_score(y_test, y_pred)
        recall = recall_score(y_test, y_pred)
        f1 = f1_score(y_test, y_pred)
        roc_auc = roc_auc_score(y_test, y_proba)

        selected_mask = best_model.named_steps["feature_selection"].get_support()
        selected_features = X_final.columns[selected_mask].tolist()

        result = {
            "Model": model_name,
            "Outer_Fold": fold_no,
            "Accuracy": accuracy,
            "Precision": precision,
            "Recall": recall,
            "F1_Score": f1,
            "ROC_AUC": roc_auc,
            "Best_Params": grid_search.best_params_,
            "Selected_Features": selected_features
        }

        all_results.append(result)

        if model_name not in roc_data:
            roc_data[model_name] = []

        fpr, tpr, _ = roc_curve(y_test, y_proba)
        roc_data[model_name].append((fpr, tpr, roc_auc))

        print("Best Params:", grid_search.best_params_)
        print("Selected Features:", selected_features)
        print(f"Accuracy : {accuracy:.4f}")
        print(f"Precision: {precision:.4f}")
        print(f"Recall   : {recall:.4f}")
        print(f"F1 Score : {f1:.4f}")
        print(f"ROC-AUC  : {roc_auc:.4f}")

        fold_no += 1

# ============================================================
# 7) RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(all_results)

print("\nTüm Fold Sonuçları:")
display(results_df)

summary_df = results_df.groupby("Model")[[
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score",
    "ROC_AUC"
]].mean().reset_index()

print("\nOrtalama Model Performansları:")
display(summary_df)

results_df.to_csv("nested_cv_all_results.csv", index=False)
summary_df.to_csv("model_performance_summary.csv", index=False)

print("\nCSV dosyaları kaydedildi:")
print("nested_cv_all_results.csv")
print("model_performance_summary.csv")

# ============================================================
# 8) ROC CURVES
# ============================================================

plt.figure(figsize=(10, 7))

for model_name, folds in roc_data.items():

    auc_values = []

    for fpr, tpr, roc_auc in folds:
        auc_values.append(roc_auc)
        plt.plot(fpr, tpr, alpha=0.25)

    mean_auc = np.mean(auc_values)

    plt.plot([], [], label=f"{model_name} Mean AUC = {mean_auc:.4f}")

plt.plot([0, 1], [0, 1], linestyle="--")

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves - Nested Cross Validation")
plt.legend()
plt.grid()
plt.show()

# ============================================================
# 9) BEST MODEL
# ============================================================

best_row = summary_df.sort_values("ROC_AUC", ascending=False).iloc[0]

print("\n====================================================")
print("EN BAŞARILI MODEL")
print("====================================================")
print("Model:", best_row["Model"])
print(f"Accuracy : {best_row['Accuracy']:.4f}")
print(f"Precision: {best_row['Precision']:.4f}")
print(f"Recall   : {best_row['Recall']:.4f}")
print(f"F1 Score : {best_row['F1_Score']:.4f}")
print(f"ROC-AUC  : {best_row['ROC_AUC']:.4f}")

print("""
Yorum:
Bu çalışmada Nested Cross Validation kullanılmıştır.
Inner loop içinde özellik seçimi ve hiperparametre optimizasyonu yapılmıştır.
Outer loop içinde ise modelin test performansı ölçülmüştür.

SVM modeli klasik SVC yerine LinearSVC ile çalıştırılmıştır.
Bunun nedeni, 100.000 örnek üzerinde RBF kernel kullanan SVC modelinin çok uzun sürede çalışmasıdır.
LinearSVC daha hızlıdır ve büyük veri setlerinde daha uygulanabilir bir çözümdür.

ROC-AUC değeri en yüksek olan model, sınıfları ayırt etme başarısı açısından en iyi model olarak seçilmiştir.
""")
```
